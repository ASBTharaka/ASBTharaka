```python
def step_function(z):
  return 1 if z>5 else 0
def predict_movie(x):
  weights=[5,2,2]
  bias=0

  z=sum(w* xi for w, xi in zip(weights,x))+bias

  return step_function(z)
```

```python
print("should you go to the movie today? Answer the following questions with 1 or 0")

weather=int(input("Is wether good(1-yes ,0-?no)"))
company=int(input("Do you have firend to go(1-yes ,0-no)"))
proximiti=int(input("Is the thearte nerby?(1-yes ,0-no)"))

user_input=[weather,company,proximiti]

desicion=predict_movie(user_input)

if desicion ==1:
  print("Decision:You can go movie!!!!")
else:
  print("Decision:You can't go movie...Stay Home!!!!")
```
